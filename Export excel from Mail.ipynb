## Extract excel file from Outlook using Power Automate

This is a simple method to extract and configure power automate to extract file from mail to sharepoint with specific **Subject** conditions

![image_1777050662459.png](./image_1777050662459.png "image_1777050662459.png")

First of all create new **Outlook task**, login with your credentials and set:
1) _Include Attachments_ = **Yes**
2) _Only with attahments_ = **Yes**
3) _Folder_ = **inbox**

![image_1777050694987.png](./image_1777050694987.png "image_1777050694987.png")


Add **Condition task** and setup with the following parameters:

![image_1777050813682.png](./image_1777050813682.png "image_1777050813682.png")

Create file on the sharepoint task and set your **Company Site Address** and **Folder Path** to locate your files. 
Configure your **File Name** as you prefer, if you want timestamp on the name you can use the following expression: 

```sql
concat('Import_',formatDateTime(utcNow(),'yyyMMdd'),'.xlsx')
```
